# Outreach Writer - local LLM Cold Outreach Generator

This notebook covers prompt engineering and generation strategies for Agent 4 (The Networker). The goal is to draft personalized, non-generic LinkedIn invitations (under 300 characters) and follow-up cold emails tailored to a candidate's portfolio projects.

```python
import os
from transformers import AutoTokenizer, AutoModelForCausalLM
```

## 1. Prompt Design for Outreach Generation

We enforce formatting limits (LinkedIn's 300-character max) and style controls (avoiding clichés like 'hope this finds you well') directly in the prompt instructions.

```python
system_prompt = """
You are an elite networking coach. Based on the target company and matching project, draft:
1. A LinkedIn connection note (strict limit of 300 characters maximum, including spaces. Reference a matching skill/project. Be direct).
2. A follow-up cold email subject line and body (conversational, low-friction call-to-action, under 150 words).
Do NOT use standard cliches like 'I hope this finds you well' or 'I am reaching out to'.
"""

user_input = """
Candidate: Alice
Target Company: Google
Target Role: Python Backend Developer
Matching Project: VinoMetrix (FastAPI/Postgres backend prediction module)
"""

prompt = f"<|system|>\n{system_prompt}\n<|user|>\n{user_input}\n<|assistant|>\n"
```

## 2. Local Model Inference Setup

For communication agents, a slightly higher temperature (e.g. `temperature=0.7`) is used to encourage natural-sounding variations.

```python
model_id = "microsoft/Phi-3-mini-4k-instruct"
try:
    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")
    
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=300, temperature=0.7, do_sample=True)
    result = tokenizer.decode(outputs[0], skip_special_tokens=True)
    print("\n--- Generated Outreach Drafts ---\n")
    print(result.replace(prompt, ""))
except Exception as e:
    print(f"Skipping actual model load during demo. Details: {e}")
```
